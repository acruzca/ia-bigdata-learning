# 01 - Introducción a NumPy

## Objetivo

Aprender los fundamentos de NumPy para trabajar con datos numéricos mediante arrays y operaciones vectorizadas.

## Conceptos utilizados

- Importación de NumPy
- Arrays (`ndarray`)
- Acceso mediante índices
- Slicing
- Operaciones vectorizadas
- Funciones estadísticas básicas
- Operaciones entre arrays
- Máscaras booleanas
- Filtrado de datos

```python
import numpy as np

# Creamos un array de precios
precios = np.array([850, 220, 45, 25, 350])

print(precios)
print(type(precios))
```

## 1. Acceso a los elementos de un array

Los elementos de un array pueden consultarse mediante sus índices.

Al igual que en las listas de Python, el primer elemento tiene índice `0` y podemos utilizar índices negativos para acceder desde el final.

```python
print("Primer elemento:", precios[0])
print("Último elemento:", precios[-1])
print("Elementos del índice 1 al 3:", precios[1:4])
```

## 2. Operaciones vectorizadas

NumPy permite realizar una operación sobre todos los elementos de un array sin necesidad de recorrerlos mediante un bucle `for`.

Esto se conoce como **vectorización**.

```python
precios_con_iva = precios * 1.21

print("Precios originales:", precios)
print("Precios con IVA:", precios_con_iva)
```

## 3. Operaciones estadísticas básicas

NumPy proporciona funciones para realizar cálculos estadísticos directamente sobre los arrays.

```python
print("Suma:", np.sum(precios))
print("Media:", np.mean(precios))
print("Máximo:", np.max(precios))
print("Mínimo:", np.min(precios))
print("Desviación típica:", np.std(precios))
```

## 4. Operaciones entre arrays

Si dos arrays tienen dimensiones compatibles, NumPy permite realizar operaciones elemento a elemento.

En este ejemplo utilizamos un array con los precios y otro con el stock para calcular el valor de cada producto.

```python
stock = np.array([4, 8, 15, 0, 3])

valor_productos = precios * stock

print("Precios:", precios)
print("Stock:", stock)
print("Valor por producto:", valor_productos)
print("Valor total del inventario:", np.sum(valor_productos), "€")
```

## 5. Ejercicio: análisis de ventas

Partimos de un array que contiene las ventas realizadas durante siete periodos.

Calcularemos:

1. Total de ventas.
2. Venta media.
3. Venta máxima.
4. Venta mínima.
5. Ventas superiores a 1.300 €.
6. Número de ventas superiores a 1.300 €.
7. Ventas resultantes tras aplicar un incremento del 10 %.

```python
ventas = np.array([1200, 950, 1430, 1100, 1750, 1620, 1380])

# 1. Total de ventas
total_ventas = np.sum(ventas)
print("Total de ventas:", total_ventas, "€")

# 2. Venta media
venta_media = np.mean(ventas)
print("Venta media:", venta_media, "€")

# 3. Venta máxima
venta_maxima = np.max(ventas)
print("Venta máxima:", venta_maxima, "€")

# 4. Venta mínima
venta_minima = np.min(ventas)
print("Venta mínima:", venta_minima, "€")

# 5. Ventas superiores a 1.300 €
ventas_superiores = ventas[ventas > 1300]
print("Ventas superiores a 1.300 €:", ventas_superiores)

# 6. Número de ventas superiores a 1.300 €
numero_ventas_superiores = np.sum(ventas > 1300)
print("Número de ventas superiores a 1.300 €:", numero_ventas_superiores)

# 7. Aumento del 10 %
ventas_aumentadas = ventas * 1.10
print("Ventas con aumento del 10 %:", ventas_aumentadas)
```

## 6. Máscaras booleanas

Una máscara booleana permite aplicar una condición a cada elemento de un array.

Por ejemplo:

`ventas > 1300`

genera un array de valores `True` y `False`.

Podemos utilizar posteriormente esa máscara para seleccionar únicamente los elementos que cumplen la condición.

```python
mascara = ventas > 1300

print("Ventas:", ventas)
print("Máscara:", mascara)
print("Ventas superiores a 1.300 €:", ventas[mascara])
```
