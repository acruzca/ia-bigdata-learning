```python
# 1. Funcion que devuelve la suma de dos numeros.
def sumar(a,b):
  return a + b

# 2. Funcion que devuleve true si la edad es mayor de 18 años.
def es_mayor_edad(edad):
  return edad >= 18

# 3. Recibe una lista y devuelve su media.
def calcular_media(numeros):
  return sum(numeros) / len(numeros)

# 4. Convertir grados celsius en fahrenheit F=(C x 9/5) + 32
def convertir_celsius_fahrenheit(celsius):
  return (celsius * 9/5) +32


# Salida de resultados
print(sumar(5,3))
print(es_mayor_edad(20))
print(es_mayor_edad(16))

datos = [10,20,30,40,50]
print(calcular_media(datos))

print(convertir_celsius_fahrenheit(25))
```
