```python
ventas = [
    {"producto": "Portátil", "cantidad": 2, "precio": 800},
    {"producto": "Monitor", "cantidad": 5, "precio": 200},
    {"producto": "Teclado", "cantidad": 10, "precio": 50},
    {"producto": "Ratón", "cantidad": 8, "precio": 30}
]

def calcular_facturacion(venta):
  return venta["cantidad"] * venta["precio"]

facturacion_total = 0

for venta in ventas:
  facturacion = calcular_facturacion(venta)

  print(
      venta["producto"],
      "->",
      facturacion,
      "€"
  )

  facturacion_total += facturacion

print("Facturación total:", facturacion_total, "€")
```
