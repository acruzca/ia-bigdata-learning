```python
productos = [
    {
      "nombre": "Portátil" ,
      "precio": 800,
      "stock": 5
    },
    {
      "nombre": "Monitor" ,
      "precio": 200,
      "stock": 10
    },
    {
      "nombre": "Teclado" ,
      "precio": 50,
      "stock": 20
    },
    {
      "nombre": "Ratón" ,
      "precio": 30,
      "stock": 15
    }

]

# 1. Mostrar todos los nombres
print("Productos:")

for producto in productos:
  print(producto["nombre"])

# 2. Mostrar productos con precio superior a 100 €
print("\nProductos con precio superior a 100 €:")

for producto in productos:
  if producto["precio"] > 100:
    print(producto['nombre'])

# 3. Mostrar el valor de stock de cada producto
print("\nValor de stock por producto:")

for producto in productos:
  valor_stock = producto["precio"] * producto['stock']

  print(
        producto['nombre'],
        "->",
        valor_stock,
        "€"
  )

# 4. Calcular valor total del inventario
valor_total = 0

for producto in productos:
  valor_total += producto['precio'] * producto['stock']

print("Valor total del inventario:", valor_total, "€")

# 5. Buscar el producto más caro
producto_mas_caro = productos[0]

for producto in productos:
  if producto["precio"] > producto_mas_caro["precio"]:
    producto_mas_caro = producto

print(
    "\nProducto más caro:",
    producto_mas_caro["nombre"],
    "-",
    producto_mas_caro["precio"],
    "€"
)

# 6. Contar productos con stock superior a 10
contador = 0

for producto in productos:
  if producto["stock"] > 10:
    contador += 1

print("\nProductos con stock superior a 10:", contador)

```
