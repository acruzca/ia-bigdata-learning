# 04 - Bucles en Python

## Objetivo

Practicar estructuras repetitivas en Python utilizando bucles `for` y `while`.

## Conceptos utilizados

- Bucle `for`
- Bucle `while`
- Contadores
- Acumuladores
- Condicionales
- Recorrido de datos

```python
# Programa que muestra los resultados sobre ventas
ventas = [1200, 850, 2300, 450, 1800]

# Venta total
total = 0

for venta in ventas:
  total += venta

# Venta media
media = total / len(ventas)

# Numero ventas
numero_ventas = len(ventas)

# Ventas superiores a 1000
ventas_superiores = []

for venta in ventas:
  if venta > 1000:
    ventas_superiores.append(venta)

# Cantidad de ventas superiores a 1000
cantidad_superiores = len(ventas_superiores)

# Mostrar resultados
print("Venta total:", total)
print("Venta media:",round(media, 2))
print("Número de ventas:", numero_ventas)
print("Ventas superiores a 1000 €:", ventas_superiores)
print("Cantidad de ventas superiores a 1000 €:", cantidad_superiores)
```
