# 01 - Fundamentos de Python

## Objetivo

Practicar los fundamentos básicos del lenguaje Python mediante el uso de variables, tipos de datos, listas y acceso a elementos.

## Conceptos utilizados

- Variables
- Números enteros
- Cadenas de texto
- Listas
- Índices
- Operaciones aritméticas
- `print()`

```python

nombre = "Antonio"
edad = 52

anio_actual = 2026
anio_nacimiento = anio_actual - edad

tecnologias = [
    "Java",
    "PHP",
    "JavaScript",
    "SQL",
    "Python"
]

print(nombre)
print(edad)
print("Naciste aproximadamente en", anio_nacimiento)

print("Tecnologias:", tecnologias)
print("Primera tecnología", tecnologias[0])
print("Ultima tecnología", tecnologias[-1])

```

## Aprendizajes

En este ejercicio he practicado la creación de variables, operaciones aritméticas sencillas y el manejo básico de listas en Python.

También he utilizado índices positivos y negativos para acceder al primer y último elemento de una lista.
