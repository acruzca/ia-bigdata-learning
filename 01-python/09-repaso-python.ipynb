# 09 - Repaso de Python: gestión de productos

## Objetivo

Realizar un ejercicio integrador para repasar los principales fundamentos de Python mediante la gestión de un inventario de productos.

## Conceptos utilizados

- Listas
- Diccionarios
- Bucles `for`
- Condicionales `if`
- Variables y acumuladores
- Funciones
- Parámetros y valores de retorno
- Búsqueda y filtrado de datos
- Operaciones aritméticas

```python
productos = [
    {"nombre": "Portátil", "precio": 850, "stock": 4},
    {"nombre": "Monitor", "precio": 220, "stock": 8},
    {"nombre": "Teclado", "precio": 45, "stock": 15},
    {"nombre": "Ratón", "precio": 25, "stock": 0},
    {"nombre": "Tablet", "precio": 350, "stock": 3}
]

# 1. Mostrar todos los productos con su precio
print("1. LISTA DE PRODUCTOS")

for producto in productos:
  print(producto["nombre"], " - ", producto["precio"], "€")

# 2. Mostrar solamente los productos con stock
print("\n2. PRODUCTOS CON STOCK")

for producto in productos:
  if producto["stock"] > 0:
    print(
      producto["nombre"],
      "- Stock:",
      producto["stock"]
  )

# 3. Calcular el valor total del inventario
print("\n3. VALOR TOTAL DEL INVENTARIO")

valor_total = 0

for producto in productos:
  valor_producto = producto["precio"] * producto["stock"]
  valor_total = valor_total + valor_producto

print("Valor total:", valor_total, "€")

# 4. Encontrar el mas caro
print("\n4. PRODUCTO MÁS CARO")

producto_mas_caro = productos[0]

for producto in productos:
  if producto["precio"] > producto_mas_caro["precio"]:
    producto_mas_caro = producto

print(
    producto_mas_caro["nombre"],
    "-",
    producto_mas_caro["precio"],
    "€"
)

#5. Calcular el precio medio
print ("\n5. PRECION MEDIO")

suma_precios = 0

for producto in productos:
  suma_precios = suma_precios + producto["precio"]

precio_medio = suma_precios / len(productos)

print("Precio medio:", precio_medio, "€")

#6. Función para buscar un producto
print("\n6. BÚSQUEDA DE PRODUCTOS")

def buscar_producto(nombre):
  for producto in productos:

    if producto["nombre"].lower() == nombre.lower():
      return producto

  return "Producto no encontrado"

print(buscar_producto("Monitor"))
print(buscar_producto("Tablet"))
print(buscar_producto("Impresora"))
```
