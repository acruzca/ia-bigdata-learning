# 02 - Listas en Python

## Objetivo

Practicar la creación, acceso y manipulación de listas en Python.

## Conceptos utilizados

- Creación de listas
- Acceso mediante índices
- Modificación de elementos
- Operaciones básicas con listas

```python
lenguajes = ["Java", "PHP", "JavaScript"]

# Añadir Python
lenguajes.insert(1, "Python")

# Eliminar PHP
lenguajes.remove("PHP")

# Añadir SQL al final
lenguajes.append("SQL")

# Mostrar resultados
print("Lista:", lenguajes)
print("Primero:", lenguajes[0])
print("Último:", lenguajes[-1])
print("Número de elementos:", len(lenguajes))
```
