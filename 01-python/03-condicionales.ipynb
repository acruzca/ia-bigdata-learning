# 03 - Condicionales en Python

## Objetivo

Practicar la toma de decisiones en Python mediante estructuras condicionales.

## Conceptos utilizados

- Condicionales `if`
- Condicionales `elif`
- Condicionales `else`
- Operadores de comparación
- Variables
- Entrada y salida de datos

```python
salario = 32000

if salario < 20000:
  print("Salario bajo")
elif salario < 30000:
  print("Salario medio")
elif salario < 50000:
  print("Salario alto")
else:
  print("Salario muy alto")
```

## Aprendizajes

En este ejercicio he practicado el uso de estructuras condicionales para ejecutar diferentes acciones en función del valor de los datos.
